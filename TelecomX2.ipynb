```python
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import OneHotEncoder
```

```python
df.info()
```

```python
x=df.drop(columns=['Churn'], axis=1)
y=df['Churn']
```

```python
colunas = x.columns
colunas
```

```python
one_hot = make_column_transformer((
    OneHotEncoder(drop='if_binary'), 
    ['Gênero','Idoso','Parceiro','Serviço de Telefone','Serviço de Internet','Linhas Múltiplas','Método de Pagamento',
     'Tipo de Contrato']
),
remainder='passthrough', 
sparse_threshold=0)
```

```python
x=one_hot.fit_transform(x)
```

```python
one_hot.get_feature_names_out(colunas)
```

```python
hotfim = pd.DataFrame(x, columns=one_hot.get_feature_names_out(colunas))
hotfim
```

```python
y.value_counts()
```

```python
import matplotlib.pyplot as plt 
y.value_counts().plot(kind='bar')
plt.title('Distribuição de Churn')
plt.xlabel('Churn')
plt.ylabel('Contagem')
```

```python

(y.value_counts(normalize=True) * 100).plot(kind='bar')

plt.title('Distribuição de Churn (%)')
plt.xlabel('Churn')
plt.ylabel('Porcentagem (%)')
plt.show()
```

Aqui vemos que a maioria dos clientes permanecem na empresa ativos, cerca de 70%, enquanto 30% se evadiram.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.boxplot(x='Churn', y='Tempo de Contrato (meses)', data=df, palette='magma')
plt.title('Distribuição do Tempo de Contrato por Evasão')
plt.legend(title='Churn', labels=['Não Evasão', 'Evasão'])
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.boxplot(x='Churn', y='Cobrança Total', data=df, palette='viridis')
plt.title('Distribuição do Total Gasto por Evasão')
plt.legend(title='Churn', labels=['Não Evasão', 'Evasão'])
plt.show()
```
