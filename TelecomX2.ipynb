```python
import pandas as pd
```

```python
df = pd.read_csv('TelecomX_Data.csv')
df
```

```python
df = df.drop(columns=['Tempo de Contrato (meses)','Cobrança Mensal','Cobrança Total','Valor Diário'], axis=1)
```

```python
df
```

```python
df = df.drop(columns=['ID Cliente'], axis=1)
```

```python
import sys
!{sys.executable} -m pip install scikit-learn
```

```python
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import OneHotEncoder
```

```python
x=df.drop(columns=['Churn'], axis=1)
y=df['Churn']
```

```python
colunas = x.columns
colunas
```

```python
one_hot = make_column_transformer((
    OneHotEncoder(drop='if_binary'), 
    ['Gênero','Idoso','Parceiro','Serviço de Telefone','Serviço de Internet','Linhas Múltiplas','Método de Pagamento',
     'Tipo de Contrato']
),
remainder='passthrough', 
sparse_threshold=0)
```

```python
x=one_hot.fit_transform(x)
```

```python
one_hot.get_feature_names_out(colunas)
```

```python
pd.DataFrame(x, columns=one_hot.get_feature_names_out(colunas))
```

```python
y.value_counts()
```

```python
import matplotlib.pyplot as plt 
y.value_counts().plot(kind='bar')
plt.title('Distribuição de Churn')
plt.xlabel('Churn')
plt.ylabel('Contagem')
```

```python

(y.value_counts(normalize=True) * 100).plot(kind='bar')

plt.title('Distribuição de Churn (%)')
plt.xlabel('Churn')
plt.ylabel('Porcentagem (%)')
plt.show()
```
